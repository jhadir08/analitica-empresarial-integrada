# Respuestas de la Actividad 1

| Concepto | Definición con mis propias palabras |
|---|---|
| Analítica empresarial integrada | Uso coordinado de datos, métodos y personas para apoyar decisiones de negocio de manera trazable. |
| Dato como activo estratégico | Información que puede inventariarse, generar valor y perder utilidad si no se mantiene o utiliza. |
| Analítica descriptiva | Resume qué ocurrió mediante conteos, totales, promedios y tendencias observadas. |
| Analítica diagnóstica | Examina relaciones y posibles factores asociados para entender por qué ocurrió un resultado. |
| Analítica predictiva | Usa patrones históricos y modelos para estimar qué podría ocurrir después. |
| Analítica prescriptiva | Compara alternativas y restricciones para recomendar qué acción conviene tomar. |
| Modelo DELTA | Marco que revisa datos, empresa, liderazgo, objetivos y analistas como condiciones para competir con analítica. |
| Diferencia entre tener datos y tener un activo de datos | Tener registros no garantiza valor: un activo requiere calidad, integración, gobierno, acceso y uso en decisiones. |

````python
import base64, io, tarfile, pathlib

DATOS = pathlib.Path("datos"); DATOS.mkdir(exist_ok=True)
with tarfile.open(fileobj=io.BytesIO(base64.b64decode(_BLOB)), mode="r:gz") as tar:
    tar.extractall(DATOS)

embarques = pd.read_csv(DATOS / "embarques.csv", parse_dates=["fecha"])
fundos    = pd.read_csv(DATOS / "fundos.csv", dtype={"ubigeo": str})
productos = pd.read_csv(DATOS / "productos.csv")
clientes  = pd.read_csv(DATOS / "clientes.csv", parse_dates=["fecha_alta"])

# Un valor vacío en grupo_experimento significa que el embarque no participó
# en una prueba. Se registra como categoría explícita; no se confunde con control.
sin_experimento = embarques["grupo_experimento"].isna().sum()
embarques["grupo_experimento"] = (
    embarques["grupo_experimento"].fillna("sin_experimento")
)
print(f"Embarques clasificados como sin_experimento: {sin_experimento:,}")

URL = str(DATOS)          # las celdas siguientes leen desde aquí

print("embarques:", embarques.shape)
print("fundos   :", fundos.shape)
print("productos:", productos.shape)
print("clientes :", clientes.shape)
print("\nDatos cargados. No hizo falta descargar nada.")
````

### Respuesta 3

No se puede calcular la utilidad de la campaña de arándano 2025 con estas columnas. El notebook contiene ingresos FOB, kilos, precios, rechazos y datos operativos, pero no contiene costos de producción, jornales, fertilizantes, empaque, transporte, seguros, comisiones, gastos financieros, impuestos ni una regla para asignar costos indirectos a la campaña. La limitación es principalmente de cobertura y gobierno del dato, no de capacidad de cálculo.


````python
NORMALIZAR_PAIS = {
    "EE.UU.": "Estados Unidos",
    "USA": "Estados Unidos",
    "ESTADOS UNIDOS": "Estados Unidos",
    "Estados unidos": "Estados Unidos",
    "Paises Bajos": "Países Bajos",
    "Holanda": "Países Bajos",
    "PAISES BAJOS": "Países Bajos",
    "UK": "Reino Unido",
    "REINO UNIDO": "Reino Unido",
}

# Conservamos `embarques` para la auditoría y analizamos una copia limpia.
emb = embarques.loc[~embarques.duplicated(keep="first")].copy()
emb["pais_destino"] = (
    emb["pais_destino"].astype(str).str.strip().replace(NORMALIZAR_PAIS).astype(object)
)

assert emb["pais_destino"].notna().all(), "Hay destinos sin valor después de normalizar."

print(f"Filas para análisis después de quitar duplicados: {len(emb):,}")
print("Mercados antes :", embarques["pais_destino"].nunique())
print("Mercados después:", emb["pais_destino"].nunique())
print(sorted(emb["pais_destino"].unique()))
````

````python
# FOB exportado por año
por_anio = (emb.groupby(emb["fecha"].dt.year)["fob_total_usd"]
              .sum()
              .div(1e6)
              .round(1)
              .rename("FOB (millones US$)"))
por_anio
````

````python
# El mismo cálculo en SQL, sin base de datos: DuckDB consulta el DataFrame directamente
duckdb.sql("""
    SELECT  pais_destino                         AS mercado,
            ROUND(SUM(fob_total_usd)/1e6, 2)     AS fob_millones_usd,
            COUNT(*)                             AS embarques
    FROM    emb
    GROUP BY mercado
    ORDER BY fob_millones_usd DESC
""").df()
````

**Pregunta 4.** El primer trimestre de 2024 cayó fuerte frente a 2023 y 2025 en la zona norte. **¿Cayó porque vendimos menos kilos o porque los vendimos más baratos?** Antes de responder, ejecuta la celda siguiente.

````python
# Precio FOB promedio ponderado = FOB total / kilos totales.
q1 = norte.loc[norte["fecha"].dt.quarter == 1].copy()
q1["anio"] = q1["fecha"].dt.year

resumen_norte_q1 = (
    q1.groupby("anio", as_index=True)
    .agg(
        toneladas=("kg_netos", lambda serie: serie.sum() / 1000),
        fob_musd=("fob_total_usd", lambda serie: serie.sum() / 1e6),
        kg_netos=("kg_netos", "sum"),
        fob_usd=("fob_total_usd", "sum"),
    )
    .assign(precio_promedio=lambda tabla: tabla["fob_usd"] / tabla["kg_netos"])
    .drop(columns=["kg_netos", "fob_usd"])
    .round({"toneladas": 0, "fob_musd": 1, "precio_promedio": 2})
)

primer_trim = resumen_norte_q1[["toneladas", "fob_musd"]]
precio_promedio = resumen_norte_q1[["precio_promedio"]]

print("Volumen y FOB de la zona norte en el primer trimestre:")
display(primer_trim)
print("Precio FOB promedio ponderado (US$/kg):")
display(precio_promedio)
````

````python
import pandas as pd, numpy as np

SERIES = {"PN01209PM": "tipo_cambio"}
FUENTE = None

try:
    tc = BCRP.get_data(SERIES, "2022-01", "2026-06")
    tc.index = pd.period_range("2022-01", periods=len(tc), freq="M").to_timestamp()
    tc = tc.rename(columns={tc.columns[0]: "tipo_cambio"})
    FUENTE = "BCRP (datos oficiales en vivo)"
except Exception as e:
    print("⚠️  No se pudo conectar al BCRP:", type(e).__name__)
    print("⚠️  ATENCIÓN: se usarán valores ILUSTRATIVOS, NO OFICIALES.")
    print("⚠️  Sirven para practicar la técnica, no para citar en un informe.\n")
    meses = pd.date_range("2022-01-01", "2026-06-01", freq="MS")
    base = np.linspace(3.84, 3.58, len(meses))
    tc = pd.DataFrame({"tipo_cambio": np.round(base + np.sin(np.arange(len(meses))/5)*0.06, 4)},
                      index=meses)
    FUENTE = "VALORES ILUSTRATIVOS — no citar como oficiales"

print("Fuente:", FUENTE)
tc.tail(6)
````

````python
# FOB mensual de AgroAndes en dólares
mensual = (emb.set_index("fecha")["fob_total_usd"]
              .resample("MS").sum()
              .to_frame("fob_usd"))

# Cruzamos con el tipo de cambio del BCRP
mensual = mensual.join(tc, how="left")
mensual["fob_soles"] = mensual["fob_usd"] * mensual["tipo_cambio"]

anual = mensual.resample("YS").agg(
    fob_musd=("fob_usd", lambda s: round(s.sum()/1e6, 1)),
    fob_msoles=("fob_soles", lambda s: round(s.sum()/1e6, 1)),
    tc_promedio=("tipo_cambio", lambda s: round(s.mean(), 3)),
)
anual.index = anual.index.year
anual
````


### BLOQUE 4 — El dato en movimiento


### 20 minutos

Davenport dice que el dato crea valor **cuando está en movimiento**, no cuando está guardado. Eso tiene una consecuencia técnica muy concreta: **si tu consulta tarda diez minutos, nadie la va a usar para decidir**.

Vamos a comprobarlo midiendo el tiempo de la misma operación con dos librerías distintas.

````python
import time

# Comparamos la misma operación con el tamaño original y con una tabla ampliada.
ruta = f"{URL}/embarques.csv"
pdf = pd.read_csv(ruta)
pldf = pl.read_csv(ruta)


def operacion_pandas(tabla):
    return (
        tabla.groupby(["pais_destino", "canal", "incoterm"])
        .agg(
            fob=("fob_total_usd", "sum"),
            kg=("kg_netos", "sum"),
            precio=("precio_fob_usd_kg", "mean"),
            n=("embarque_id", "count"),
        )
        .sort_values("fob")
    )


def operacion_polars(tabla):
    return (
        tabla.group_by(["pais_destino", "canal", "incoterm"])
        .agg(
            pl.col("fob_total_usd").sum().alias("fob"),
            pl.col("kg_netos").sum().alias("kg"),
            pl.col("precio_fob_usd_kg").mean().alias("precio"),
            pl.len().alias("n"),
        )
        .sort("fob")
    )


def cronometrar(funcion, repeticiones=3):
    funcion()  # calentamiento
    tiempos = []
    for _ in range(repeticiones):
        inicio = time.perf_counter()
        funcion()
        tiempos.append(time.perf_counter() - inicio)
    return min(tiempos) * 1000


pdf_g = pd.concat([pdf] * 100, ignore_index=True)
pldf_g = pl.concat([pldf] * 100)
resultados_benchmark = []

for etiqueta, pandas_tabla, polars_tabla in [
    ("Original — 15 705 filas", pdf, pldf),
    (f"Replicado — {len(pdf_g):,} filas", pdf_g, pldf_g),
]:
    pandas_ms = cronometrar(lambda: operacion_pandas(pandas_tabla))
    polars_ms = cronometrar(lambda: operacion_polars(polars_tabla))
    resultados_benchmark.append(
        {
            "tamaño": etiqueta,
            "pandas_ms": round(pandas_ms, 1),
            "polars_ms": round(polars_ms, 1),
            "más rápido": "Polars" if polars_ms < pandas_ms else "pandas",
            "diferencia_x": round(max(pandas_ms, polars_ms) / min(pandas_ms, polars_ms), 2),
        }
    )

benchmark = pd.DataFrame(resultados_benchmark)
display(benchmark)
````


### Pregunta 6a — Mide, no supongas

Antes de ejecutar la celda anterior, casi todos asumen que Polars gana siempre. **Anota el resultado que te salió a ti:** pandas ___ ms, Polars ___ ms. Ahora ejecuta la celda de nuevo pero **sin la replicación** (con las 15 705 filas originales). ¿Sigue ganando el mismo? ¿Por qué crees que cambia?

Ésta es la primera lección de rigor del curso: **una afirmación sobre rendimiento sin medición es una opinión**. Y las mediciones dependen del tamaño del dato, de la máquina y de la operación. Un analista serio reporta las tres cosas.
Sobre 1.6 millones de filas la diferencia ya se nota. Ése es el orden de magnitud real de una agroexportadora grande del Perú: Camposol, Danper o Virú registran millones de movimientos al año entre embarques, lotes, jabas y controles de calidad. Ahí la diferencia entre segundos y minutos decide algo muy concreto: si el gerente comercial consulta el dato **antes** de la reunión, o decide sin él. Eso es «el dato en movimiento».

### Respuesta 6a

La medición depende del tamaño de la tabla, la máquina y la operación; por eso no se debe afirmar que una librería siempre es más rápida. La tabla `benchmark` muestra los milisegundos y la diferencia relativa para las 15 705 filas originales y para la tabla replicada. En la ejecución verificada, Polars fue más rápido en ambos tamaños. Al cambiar el tamaño o el entorno, la diferencia puede reducirse o cambiar por el costo fijo de preparar la operación; por eso siempre se deben reportar los tiempos observados y las condiciones de la prueba.


````python
# Polars también evalúa de forma perezosa: arma el plan y solo lee lo necesario
consulta = (pl.scan_csv(ruta)
              .filter(pl.col("canal") == "Retail")
              .group_by("pais_destino")
              .agg(pl.col("fob_total_usd").sum().alias("fob"))
              .sort("fob", descending=True))

print(consulta.explain())    # el plan de ejecución, antes de tocar un solo dato
````

````python
# Radar de madurez analítica
etiquetas = list(delta.keys())
valores = pd.to_numeric(
    pd.Series([delta[k]["puntaje"] for k in etiquetas]), errors="coerce"
).tolist()

if any(pd.isna(valor) or not 1 <= valor <= 5 for valor in valores):
    print("Complete los cinco puntajes DELTA con valores del 1 al 5 para generar el radar.")
else:
    fig = go.Figure()
    fig.add_trace(go.Scatterpolar(
        r=valores + [valores[0]],
        theta=etiquetas + [etiquetas[0]],
        fill="toself",
        name="AgroAndes Export S.A.C.",
        line=dict(color="#1E3A5F", width=2),
        fillcolor="rgba(30,58,95,0.25)",
    ))
    fig.add_trace(go.Scatterpolar(
        r=[4] * (len(etiquetas) + 1),
        theta=etiquetas + [etiquetas[0]],
        name="Nivel objetivo",
        line=dict(color="#B71C1C", width=1, dash="dash"),
    ))
    fig.update_layout(
        title="Madurez analítica — modelo DELTA",
        polar=dict(radialaxis=dict(visible=True, range=[0, 5], dtick=1)),
        showlegend=True, height=520,
    )
    fig.show()
````

## Actividad 3 — Informe para la gerencia

**A.** AgroAndes opera principalmente en los escalones descriptivo y diagnóstico inicial: puede calcular FOB, toneladas, mercados y precios, y relacionar la caída del primer trimestre de 2024 con el volumen. **B.** Corregiría primero las 120 filas duplicadas porque pueden inflar simultáneamente el número de embarques, los kilos y el FOB acumulado. Al eliminarlas, la base pasa de 15 705 a 15 585 registros, dejando una base más confiable para los demás indicadores.

**C.** Es principalmente un problema organizacional y de gobierno del dato, aunque también requiere soporte tecnológico. La empresa registra ingresos y operaciones, pero no integra los costos ni define responsables y reglas para asignarlos a una campaña; por eso no puede calcular utilidad aunque tenga herramientas de análisis.

**D.** Recomiendo: (1) aplicar una clave única y controles automáticos contra duplicados; (2) validar que cada embarque tenga precio FOB y un número de cajas mayor que cero; y (3) normalizar mercados y crear un tablero mensual para Piura y Lambayeque con toneladas, FOB/kg, rechazos y causas documentadas.

**E.** Con presupuesto limitado invertiría primero en Liderazgo, la dimensión con menor puntaje. En un año esperaría una rutina formal de revisión, responsables por KPI y decisiones registradas con evidencia; el resultado debería ser que las variaciones relevantes tengan una causa investigada, una acción asignada y seguimiento de su efecto.

## Conclusiones

- La calidad de los datos condiciona las decisiones: 189 precios nulos, 120 duplicados, 60 embarques con cero cajas y 17 nombres para 8 mercados requieren controles antes de reportar.
- En la zona norte, el primer trimestre de 2024 presentó 1 695 toneladas y US$ 3,17/kg; la caída observada se asocia al volumen y no a un precio FOB menor.
